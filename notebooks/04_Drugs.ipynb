### Top 20 des prescriptions : 

```python
top_prescriptions = sql("""
SELECT
    c.concept_name,
    COUNT(*) nb
FROM omop.drug_exposure d
JOIN omop.concept c
ON d.drug_concept_id=c.concept_id
GROUP BY c.concept_name
ORDER BY nb DESC
LIMIT 20;
""")

top_prescriptions
```

### Nombre de prescriptions par patient : 

```python
prescriptions_per_patient = sql("""
SELECT
    person_id,
    COUNT(*) nb
FROM omop.drug_exposure
GROUP BY person_id;
""")
```

```python
plt.hist(prescriptions_per_patient["nb"])
```

```python
ppp_ordered = prescriptions_per_patient.sort_values(by=["nb"], ascending=False)
ppp_ordered.head()
```

On retrouve quelques valeurs extrêmes : un patient a 361 prescriptions, un autre en a 483, puis un autre 553. A  noter que le max est pour le patient 5858 qui possédait 321 séjours.
